```python
import os 
import datetime
from langchain_core.documents import Document


def crawling(url):
    os.makedirs("crawled", exist_ok=True)
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    path = f"crawled/output_{timestamp}.txt"
    
    # !crawl4ai run --url "{url}" --output markdown --path "{path}"
    !crwl crawl "{url}" -o markdown -O "{path}"


    if not os.path.exists(path):
        raise FileNotFoundError(f"crawl4ai did not create the file: {path}")
    with open(path, "r", encoding="utf-8",  errors="ignore") as f:
        raw_text = f.read()

    # Wrap into a Document
    docs = [Document(page_content=raw_text, metadata={"source": path})]
    print(docs[0].page_content[:500].encode("utf-8", errors="ignore").decode("utf-8"))
    return docs 
```
